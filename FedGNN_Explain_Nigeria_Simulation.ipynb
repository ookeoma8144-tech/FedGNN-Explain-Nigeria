```python
import torch
import numpy as np

print("--- Step 1: Filtering the Data ---")
# 1. Keep only TRANSFER and CASH_OUT where the fraud rings operate
df_filtered = df[df['type'].isin(['TRANSFER', 'CASH_OUT'])].copy()

# 2. To ensure Colab runs smoothly, we take a chronological slice of 250,000 transactions
# (This still leaves us with thousands of fraud cases to train on!)
df_graph = df_filtered.sort_values('step').head(250000).copy()

print(f"Simulation dataset reduced to: {df_graph.shape[0]:,} transactions.")
print(f"Total Fraud cases captured in this slice: {df_graph['isFraud'].sum():,}\n")
```

```python
from torch_geometric.data import Data
from sklearn.preprocessing import StandardScaler

print("--- Step 2: Building the Graph Network ---")

# 1. Create unique Node IDs for every bank account
print("Mapping bank accounts to graph nodes...")
all_accounts = list(set(df_graph['nameOrig']).union(set(df_graph['nameDest'])))
account_mapping = {account: i for i, account in enumerate(all_accounts)}

# 2. Build the Edge Index (The Connections)
# PyTorch Geometric needs a 2D array: [Source Nodes, Target Nodes]
source_nodes = df_graph['nameOrig'].map(account_mapping).values
target_nodes = df_graph['nameDest'].map(account_mapping).values

edge_index = torch.tensor([source_nodes, target_nodes], dtype=torch.long)

# 3. Build Edge Features (The Transaction Details)
# We will give the GNN the transaction 'amount' to help it learn
scaler = StandardScaler()
amounts_scaled = scaler.fit_transform(df_graph[['amount']])
edge_attr = torch.tensor(amounts_scaled, dtype=torch.float)

# 4. Build the Labels (What we want to predict)
y = torch.tensor(df_graph['isFraud'].values, dtype=torch.float)

# 5. Compile the official PyTorch Geometric Data Object
# (We give every node a dummy feature of '1' just to initialize the network)
num_nodes = len(all_accounts)
x = torch.ones((num_nodes, 1), dtype=torch.float)

graph_data = Data(x=x, edge_index=edge_index, edge_attr=edge_attr, y=y)

print("\nSuccess! The Graph Object is ready:")
print(f"Total Nodes (Bank Accounts): {graph_data.num_nodes:,}")
print(f"Total Edges (Transactions): {graph_data.num_edges:,}")
```

```python
import torch

print("--- Step 3 (Updated): Simulating a 3-Bank Federated Environment ---")

# 1. Get the total number of transactions (edges)
total_edges = graph_data.num_edges

# 2. Shuffle the transactions to ensure a random distribution of fraud
indices = torch.randperm(total_edges)

# 3. Calculate the split points for 3 banks
one_third = total_edges // 3
two_thirds = one_third * 2

bank_a_indices = indices[:one_third]
bank_b_indices = indices[one_third:two_thirds]
bank_c_indices = indices[two_thirds:]

# 4. Create Bank A's private graph (e.g., OPay)
edge_index_A = graph_data.edge_index[:, bank_a_indices]
edge_attr_A = graph_data.edge_attr[bank_a_indices]
y_A = graph_data.y[bank_a_indices]

# 5. Create Bank B's private graph (e.g., GTBank)
edge_index_B = graph_data.edge_index[:, bank_b_indices]
edge_attr_B = graph_data.edge_attr[bank_b_indices]
y_B = graph_data.y[bank_b_indices]

# 6. Create Bank C's private graph (e.g., PalmPay)
edge_index_C = graph_data.edge_index[:, bank_c_indices]
edge_attr_C = graph_data.edge_attr[bank_c_indices]
y_C = graph_data.y[bank_c_indices]

print("3-Way Federated Split Successful!")
print(f"Bank A holds {edge_index_A.size(1):,} transactions securely.")
print(f"Bank B holds {edge_index_B.size(1):,} transactions securely.")
print(f"Bank C holds {edge_index_C.size(1):,} transactions securely.")
```

```python
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv
import torch.nn as nn

print("--- Step 4: Initializing the GNN Architecture ---")

class FedFraudGNN(torch.nn.Module):
    def __init__(self, hidden_channels):
        super(FedFraudGNN, self).__init__()

        # 1. Message Passing Layers (Learning the network of accounts)
        # SAGEConv is excellent for scaling to massive banking datasets
        self.conv1 = SAGEConv((-1, -1), hidden_channels)
        self.conv2 = SAGEConv((-1, -1), hidden_channels)

        # 2. The Fraud Classifier (Scoring the transaction)
        self.classifier = nn.Sequential(
            nn.Linear(hidden_channels * 2, hidden_channels),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(hidden_channels, 1) # Outputs a single fraud risk score
        )

    def forward(self, x, edge_index):
        # Step A: Generate intelligent profiles for every bank account (Nodes)
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = F.dropout(x, p=0.5, training=self.training)
        x = self.conv2(x, edge_index)

        # Step B: Look at the specific transactions (Edges)
        src_nodes = edge_index[0] # The Sender
        dst_nodes = edge_index[1] # The Receiver

        # Combine the sender and receiver profiles
        edge_features = torch.cat([x[src_nodes], x[dst_nodes]], dim=-1)

        # Step C: Predict if this specific transfer is part of a fraud ring
        return self.classifier(edge_features).squeeze()

print("Model architecture successfully defined!")
```

```python
import copy
import torch.optim as optim
import torch.nn as nn

print("--- Step 5: Simulating Federated Averaging (FedAvg) ---")

# 1. Initialize the Global Model (The Central Bank's Master AI)
hidden_channels = 64
global_model = FedFraudGNN(hidden_channels=hidden_channels)

# Because fraud is so rare (0.1%), we have to "weight" the loss function.
# This forces the AI to pay 500x more attention to a fraud case than a normal transaction.
pos_weight = torch.tensor([500.0])
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

# 2. Define the Local Training Function (What happens inside the bank's firewalls)
def train_local_model(local_model, node_features, local_edge_index, local_y, epochs=3):
    local_model.train()
    optimizer = optim.Adam(local_model.parameters(), lr=0.01)

    for epoch in range(epochs):
        optimizer.zero_grad()
        # The model looks at the bank's specific network of transactions
        out = local_model(node_features, local_edge_index)
        loss = criterion(out, local_y)
        loss.backward()
        optimizer.step()

    # Return the newly learned mathematical weights and the final loss
    return local_model.state_dict(), loss.item()

# 3. The Federated Communication Loop
communication_rounds = 5 # In a real paper, you might do 20-50, but 5 is great for simulation
local_epochs = 3

for round_num in range(communication_rounds):
    print(f"\n--- Communication Round {round_num + 1}/{communication_rounds} ---")

    # A. Distribute the current Global Model to the Banks
    model_A = copy.deepcopy(global_model)
    model_B = copy.deepcopy(global_model)
    model_C = copy.deepcopy(global_model)

    # B. Train Locally (Strictly on local edges/transactions)
    weights_A, loss_A = train_local_model(model_A, graph_data.x, edge_index_A, y_A, local_epochs)
    print(f"Bank A (OPay) Local Loss:   {loss_A:.4f}")

    weights_B, loss_B = train_local_model(model_B, graph_data.x, edge_index_B, y_B, local_epochs)
    print(f"Bank B (GTBank) Local Loss: {loss_B:.4f}")

    weights_C, loss_C = train_local_model(model_C, graph_data.x, edge_index_C, y_C, local_epochs)
    print(f"Bank C (PalmPay) Local Loss:{loss_C:.4f}")

    # C. The Magic of FedAvg: Combining the knowledge securely
    averaged_weights = copy.deepcopy(weights_A)
    for key in averaged_weights.keys():
        # Formula: (Weights_A + Weights_B + Weights_C) / 3
        averaged_weights[key] = (weights_A[key] + weights_B[key] + weights_C[key]) / 3.0

    # D. Update the Global Model with the new collaborative knowledge
    global_model.load_state_dict(averaged_weights)
    print("-> Global Model successfully updated with averaged knowledge.")

print("\n✅ Federated Training Complete! The Global Model is ready to detect cross-institution fraud rings.")
```
